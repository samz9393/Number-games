```python

import random

coins = 0
shields = 0


# SECRET NUMBER GAME
def secret_number_game():
    global coins

    while True:
        print("\n🧙 CHOOSE YOUR DIFFICULTY")
        print("1. Easy (1 to 10)")
        print("2. Hard (1 to 100)")
        print("3. Impossible (1 to 1000)")
        print("Type r to return to the main menu.")

        mode = input("Choose: ").lower().strip()

        if mode == "r":
            return
        elif mode == "1":
            number_range = 10
        elif mode == "2":
            number_range = 100
        elif mode == "3":
            number_range = 1000
        else:
            print("Invalid choice!")
            continue

        secret = random.randint(1, number_range)
        guesses = 0

        while True:
            answer = input(
                f"\n🧙 Guess a number from 1 to {number_range}: "
            ).lower().strip()

            if answer == "r":
                print("Returning to the main menu...")
                return

            try:
                guess = int(answer)
            except ValueError:
                print("Enter a whole number or type r!")
                continue

            if guess < 1 or guess > number_range:
                print("Choose a number within the correct range!")
                continue

            guesses += 1

            if guess == secret:
                print("\n✨ YOU GUESSED IT!")
                print("The secret number was", secret)
                print("Guesses:", guesses)

                coins += 10

                print("🪙 You earned 10 coins!")
                print("Total coins:", coins)
                break

            elif guess > secret:
                print("The secret number is LOWER!")
            else:
                print("The secret number is HIGHER!")

            if mode == "3":
                difference = abs(secret - guess)

                if difference <= 10:
                    print("Magic hint: You are extremely close!")
                elif difference <= 50:
                    print("Magic hint: You are getting closer!")
                else:
                    print("Magic hint: You are still far away!")

        again = input(
            "\nPlay Secret Number again? (yes/no/r for menu): "
        ).lower().strip()

        if again == "r":
            return
        elif again != "yes":
            return


# BOMB NUMBER GAME
def bomb_game():
    global coins, shields

    while True:
        print("\n💣 BOMB NUMBER")
        print("Choose safe numbers from 1 to 20!")
        print("1. Easy - 1 bomb (1 coin per safe pick)")
        print("2. Medium - 2 bombs (2 coins per safe pick)")
        print("3. Hard - 3 bombs (3 coins per safe pick)")
        print("Type r to return to the main menu.")

        mode = input("Choose difficulty: ").lower().strip()

        if mode == "r":
            return
        elif mode == "1":
            bomb_count = 1
            reward = 1
        elif mode == "2":
            bomb_count = 2
            reward = 2
        elif mode == "3":
            bomb_count = 3
            reward = 3
        else:
            print("Invalid choice!")
            continue

        starting_coins = coins
        round_coins = 0

        bombs = set(random.sample(range(1, 21), bomb_count))
        chosen = set()
        game_over = False

        print("\n💣 THE BOMBS ARE HIDDEN!")
        print("You have", shields, "shield(s).")
        print("Safe picks earn coins.")
        print("Hitting a bomb without a shield loses")
        print("all coins earned during this round.")
        print("Type r at any time to return to the menu.")

        while True:
            print("\n📋 NUMBERS ALREADY PICKED:",
                  sorted(chosen) if chosen else "None")

            answer = input(
                "\nChoose a number (1-20): "
            ).lower().strip()

            if answer == "r":
                print("Returning to the main menu...")
                print("You keep the coins earned so far.")
                return

            try:
                number = int(answer)
            except ValueError:
                print("Enter a whole number or type r!")
                continue

            if number < 1 or number > 20:
                print("Choose a number from 1 to 20!")
                continue

            if number in chosen:
                print("You already picked that number!")
                continue

            chosen.add(number)

            if number in bombs:

                if shields > 0:
                    shields -= 1
                    bombs.remove(number)

                    print("\n🛡️ YOUR SHIELD PROTECTED YOU!")
                    print("The bomb has been disarmed.")
                    print("Shields left:", shields)

                    # Replace the bomb with a new hidden bomb
                    available_numbers = [
                        n for n in range(1, 21)
                        if n not in chosen and n not in bombs
                    ]

                    if available_numbers:
                        new_bomb = random.choice(available_numbers)
                        bombs.add(new_bomb)

                        print("💣 A NEW BOMB HAS APPEARED!")
                        print("It is hiding behind a different number.")

                else:
                    print("\n💥 BOOM! YOU HIT A BOMB!")
                    print("You lost", round_coins,
                          "coins earned during this round!")

                    coins = starting_coins
                    round_coins = 0

                    print("Your coins are now:", coins)
                    print("GAME OVER!")

                    game_over = True
                    break

            else:
                coins += reward
                round_coins += reward

                print("\n✨ SAFE PICK!")
                print("You earned", reward, "coins!")
                print("Coins earned this round:", round_coins)
                print("🪙 Total coins:", coins)

            # Win after choosing all safe numbers
            if len(chosen) == 20 - len(bombs):
                break

        if not game_over:
            print("\n🎉 YOU WON BOMB NUMBER!")
            coins += 10
            print("Bonus reward: 10 coins!")
            print("🪙 Total coins:", coins)

        again = input(
            "\nPlay Bomb Number again? (yes/no/r for menu): "
        ).lower().strip()

        if again == "r":
            return
        elif again != "yes":
            return


# COIN SHOP
def shop():
    global coins, shields

    while True:
        print("\n" + "=" * 35)
        print("          🏪 WIZARD SHOP")
        print("=" * 35)

        print("🪙 Coins:", coins)
        print("🛡️ Shields:", shields)

        print("\n1. Buy a Bomb Shield - 10 coins")
        print("2. Leave the shop")

        choice = input("Choose an option: ").strip()

        if choice == "1":

            if coins >= 10:
                coins -= 10
                shields += 1

                print("\n🛡️ YOU BOUGHT A SHIELD!")
                print("Shields:", shields)
                print("Coins left:", coins)

            else:
                print("Not enough coins!")
                print("Play games to earn more.")

        elif choice == "2":
            return

        else:
            print("Invalid choice!")


# HOW TO PLAY
def how_to_play():
    print("\n📖 HOW TO PLAY")

    print("\n🎯 SECRET NUMBER")
    print("- Guess the hidden number.")
    print("- Easy is 1 to 10.")
    print("- Hard is 1 to 100.")
    print("- Impossible is 1 to 1000.")
    print("- Guess correctly to earn 10 coins.")
    print("- Follow the higher and lower hints.")

    print("\n💣 BOMB NUMBER")
    print("- Choose numbers from 1 to 20.")
    print("- Avoid the hidden bombs.")
    print("- Easy has 1 bomb and gives 1 coin per safe pick.")
    print("- Medium has 2 bombs and gives 2 coins per safe pick.")
    print("- Hard has 3 bombs and gives 3 coins per safe pick.")
    print("- A shield protects you from one bomb.")
    print("- When a shield protects you, a new bomb appears")
    print("  behind a different, unpicked number.")
    print("- Hitting a bomb without a shield loses all coins")
    print("  earned during that round.")
    print("- Coins from before the round are kept.")
    print("- Winning gives a 10-coin bonus.")

    print("\n🏪 SHOP")
    print("- Buy a shield for 10 coins.")
    print("- Shields protect you from one bomb.")

    print("\n🪙 CURRENCY")
    print("- Your coins and shields are shown in the main menu.")
    print("- Your coins and shields stay saved while playing.")

    print("\n🔙 RETURN TO MENU")
    print("- Type r at any time during either game.")
    print("- Your coins and shields stay saved.")

    input("\nPress Enter to return to the menu...")


# MAIN MENU
running = True

while running:

    print("\n" + "=" * 40)
    print("       🧙 THE WIZARD'S WORLD")
    print("=" * 40)

    print("🪙 COINS:", coins)
    print("🛡️ SHIELDS:", shields)

    print("=" * 40)

    print("\n1. 🎯 Secret Number Game")
    print("2. 💣 Bomb Number Game")
    print("3. 🏪 Coin Shop")
    print("4. 📖 How to Play")
    print("5. 🚪 Quit")

    choice = input("\nChoose an option (1-5): ").strip()

    if choice == "1":
        secret_number_game()

    elif choice == "2":
        bomb_game()

    elif choice == "3":
        shop()

    elif choice == "4":
        how_to_play()

    elif choice == "5":
        print("\n🧙 THE WIZARD SAYS: FAREWELL!")
        print("🪙 Final coins:", coins)
        print("🛡️ Final shields:", shields)
        print("✨ Thanks for playing!")
        running = False

    else:
        print("Invalid choice! Choose 1, 2, 3, 4, or 5.")
```
